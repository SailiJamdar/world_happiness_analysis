```python
query="""
SELECT "Regional indicator",
      ROUND(AVG("Ladder score"),2) as avg_happiness,
      count(*) as num_countries
from happiness
group by "Regional indicator"
order by avg_happiness DESC

"""
print(pd.read_sql(query,conn))
```
